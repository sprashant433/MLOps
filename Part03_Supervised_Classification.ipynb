## 3.1 Logistic Regression

**Formula:** `P(y=1) = 1 / (1 + e^-(β₀ + β₁x₁ + ...))`

**Use Case:** Binary and multi-class classification, interpretable probabilities

**Output:** Probabilities between 0 and 1

### 📌 Logistic Regression — Clear + Technical Explanation

Logistic Regression is a **classification algorithm** (not regression despite the name) used to predict **probabilities** of classes (usually binary: 0/1).

👉 Instead of predicting a continuous value (like linear regression), it:

* Predicts a **probability (0 to 1)**
* Uses a **sigmoid function** to squash outputs

---

### 🔢 Step 1: Linear Combination

Like linear regression:

$$
z = w_1 x_1 + w_2 x_2 + \cdots + b
$$

---

### 🔢 Step 2: Sigmoid Function

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

👉 Converts output into probability:

* Output close to 1 → Class 1
* Output close to 0 → Class 0

---

### 🎯 Decision Rule

```text
if probability ≥ 0.5 → Class 1
else → Class 0
```

---

### 🔢 Loss Function (Very Important)

Logistic Regression uses **Log Loss (Cross-Entropy)**:
$$
Loss = -[y \log(p) + (1-y)\log(1-p)]
$$
👉 Penalizes wrong predictions heavily

---

### 📊 Example

| Study Hours | Pass (y) |
| ----------- | -------- |
| 1           | 0        |
| 2           | 0        |
| 4           | 1        |
| 6           | 1        |

👉 Model learns:

* More hours → higher probability of passing

---

### 🚀 Python Example

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

X = np.array([[1], [2], [4], [6]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

print(model.predict([[3]]))        # class
print(model.predict_proba([[3]]))  # probability
```

---

### ⚙️ Important Concepts

#### 🔹 1. Log-Odds (Logit)
$$
\log\left(\frac{p}{1-p}\right)
$$
👉 Logistic regression is actually **linear in log-odds**

---

#### 🔹 2. Decision Boundary

* Linear (straight line / hyperplane)
* Separates classes

---

#### 🔹 3. Regularization

* L1 (Lasso) → feature selection
* L2 (Ridge) → coefficient shrinkage

---

### ⚔️ Logistic vs Linear Regression

| Feature  | Logistic          | Linear     |
| -------- | ----------------- | ---------- |
| Output   | Probability (0–1) | Continuous |
| Use case | Classification    | Regression |
| Function | Sigmoid           | Line       |

---

### ⚔️ Logistic vs SVM

| Feature          | Logistic      | SVM             |
| ---------------- | ------------- | --------------- |
| Output           | Probabilities | Margin-based    |
| Interpretability | High          | Medium          |
| Speed            | Faster        | Slower (kernel) |

---

### 🔥 Advantages

* Simple & interpretable
* Works well for linearly separable data
* Outputs probabilities

---

### ⚠️ Limitations

* Cannot capture complex non-linear patterns
* Sensitive to outliers
* Assumes linear boundary

---

### 🚀 Real-world Use Cases

* Spam detection

* Click-through rate prediction

👉 For **K classes**, train **K binary classifiers**

Example (3 classes: A, B, C):

* Model 1 → A vs (B, C)
* Model 2 → B vs (A, C)

👉 Final prediction = class with **highest probability**

---

### 🔹 2. Softmax (Multinomial Logistic Regression)

👉 Instead of multiple models, use **one model** with softmax

#### 🔢 Softmax Function
$$
P(y = k) = \frac{e^{z_k}}{\sum_{j=1}^{K} e^{z_j}}
$$
👉 Outputs probabilities for all classes:

* Sum = 1
* Picks highest probability

---

### ⚔️ OvR vs Softmax

| Feature     | OvR          | Softmax           |
| ----------- | ------------ | ----------------- |
| Models      | Multiple (K) | Single            |
| Simplicity  | Easy         | More mathematical |
| Performance | Good         | Better (usually)  |
| Use case    | Default      | Preferred in ML   |

---

👉 Idea is very simple:

> “Tell me who your neighbors are, I’ll tell you who you are.”

* Look at **K nearest points**
* Predict based on their values

---

### 🚀 How It Works (Step-by-Step)

1. Choose **K (number of neighbors)**
2. Compute distance between new point and all training points
3. Pick **K closest points**
4. Predict:

* Classification → **majority vote**
* Regression → **average value**

---

### 📊 Example (Classification)

#### Data:

| Point | Class |
| ----- | ----- |
| (1,1) | A     |
| (2,2) | A     |
| (5,5) | B     |
| (6,6) | B     |

👉 New point: (3,3), K=3

Nearest neighbors:

```
(2,2) → A  
(1,1) → A  
(5,5) → B
```

👉 Majority = **A**

---

### 📏 Distance Metrics

#### 🔹 Euclidean (most common)
$$
d = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}
$$
---

#### 🔹 Others

* Manhattan distance
* Cosine similarity

---

### ⚙️ Key Parameter: K

| K Value | Behavior                           |
| ------- | ---------------------------------- |
| Small K | Low bias, high variance (overfit)  |
| Large K | High bias, low variance (underfit) |

---

### 🚀 Python Example

```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

X = np.array([[1,1], [2,2], [5,5], [6,6]])
y = np.array([0, 0, 1, 1])

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

print(model.predict([[3,3]]))
```

---

### ⚔️ KNN vs Other Models

| Feature               | KNN        | Logistic Regression |
| --------------------- | ---------- | ------------------- |
| Training time         | Very low ✅ | Medium              |
| Prediction time       | High ❌     | Low                 |
| Interpretability      | Low        | High                |
| Handles non-linearity | Yes ✅      | No                  |

---

### 🔥 Advantages

* Simple & intuitive
* No training phase
* Works well for small datasets

---

### ⚠️ Disadvantages

* Slow for large datasets ❌
* Sensitive to noise
* Needs feature scaling ⚠️

---

### ⚡ Important Concept: Feature Scaling

👉 KNN is distance-based → scale matters

```python
from sklearn.preprocessing import StandardScaler
```

---

### 🧠 Interview Insight (VERY IMPORTANT)

👉 If interviewer asks:

**“Why scaling is important in KNN?”**

Answer:

* Because distance calculation is affected by feature magnitude

---

### 🚀 Real-world Use Cases

* Recommendation systems
* Image classification (basic)
* Pattern recognition

---

### 💡 One-liner to Remember

👉 **KNN = Predict based on nearest neighbors**

---

### ⚡ Pro Tip (MLE Level)

* Use KD-Tree / Ball Tree for optimization
* Avoid KNN in:

  * Large-scale production systems
  * High-dimensional data (curse of dimensionality)

---


````python
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

# Scale features (REQUIRED for KNN!)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train model
model = KNeighborsClassifier(
    n_neighbors=5,          # number of neighbors
    weights='uniform',      # 'uniform' or 'distance' (closer neighbors have more weight)
    metric='minkowski',     # distance metric
    p=2                     # p=2 is Euclidean, p=1 is Manhattan
)
model.fit(X_train_scaled, y_train)

# Predictions
y_pred = model.predict(X_test_scaled)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")

# Find optimal K
k_range = range(1, 31)
scores = []
for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    scores.append(knn.score(X_test_scaled, y_test))

plt.plot(k_range, scores)
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.title('KNN: Accuracy vs K')
plt.show()
print(f"Best K: {k_range[scores.index(max(scores))]}")
````

## 3.3 Support Vector Machine (SVM) - Linear

**Concept:** Find hyperplane that maximizes margin between classes

**Use Case:** Binary classification, high-dimensional data

**Requires:** Feature scaling

````python
from sklearn.svm import SVC

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train linear SVM
model = SVC(
    kernel='linear',        # linear kernel
    C=1.0,                  # regularization (smaller = more regularization)
    probability=True,       # enable probability estimates
    random_state=42
)
model.fit(X_train_scaled, y_train)

# Predictions
y_pred = model.predict(X_test_scaled)
y_proba = model.predict_proba(X_test_scaled)

print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")
print(f"Number of support vectors: {len(model.support_vectors_)}")
````

## 3.4 Kernel SVM (Non-linear)

**Concept:** Use kernel trick to transform data to higher dimensions

**Kernels:** 'rbf' (Gaussian), 'poly' (polynomial), 'sigmoid'

**Use Case:** Non-linear decision boundaries

````python
# RBF Kernel (most common)
model = SVC(
    kernel='rbf',           # radial basis function
    C=1.0,                  # regularization
    gamma='scale',          # kernel coefficient ('scale', 'auto', or float)
    probability=True,
    random_state=42
)
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
print(f"RBF Kernel Accuracy: {accuracy_score(y_test, y_pred):.3f}")

# Polynomial Kernel
model_poly = SVC(kernel='poly', degree=3, C=1.0, probability=True, random_state=42)
model_poly.fit(X_train_scaled, y_train)
y_pred_poly = model_poly.predict(X_test_scaled)
print(f"Polynomial Kernel Accuracy: {accuracy_score(y_test, y_pred_poly):.3f}")
````

### 🔥 Key Idea: Kernel Trick

👉 Instead of explicitly transforming data, we use a **kernel function**
$$
K(x_i, x_j) = \phi(x_i) \cdot \phi(x_j)
$$
👉 Computes similarity in higher dimension **without computing transformation directly**

---

### 📊 Common Kernels

#### 🔹 1. Linear Kernel
$$
K(x_i, x_j) = x_i \cdot x_j
$$
👉 Same as basic SVM

---

#### 🔹 2. Polynomial Kernel
$$
K(x_i, x_j) = (x_i \cdot x_j + c)^d
$$
👉 Captures curved relationships

---

#### 🔹 3. RBF (Gaussian) Kernel 🔥 (Most Used)

$$
K(x_i, x_j) = \exp(-\gamma \|x_i - x_j\|^2)
$$

👉 Measures similarity based on distance

---

### ⚙️ Important Parameters

#### 🔹 C (Regularization)

* High C → low bias, high variance (overfit)
* Low C → high bias, low variance

---

#### 🔹 γ (Gamma) — for RBF

* High γ → very tight boundary (overfit)
* Low γ → smoother boundary

---

### 📊 Example

👉 Data shaped like a circle:

* Linear SVM ❌ cannot separate
* Kernel SVM ✅ creates circular boundary

---

### 🚀 Python Example


### 🔢 Bayes Theorem

$$
P(C|X) = \frac{P(X|C) \cdot P(C)}{P(X)}
$$

Where:

* (P(C|X)) → Posterior (what we want)
* (P(X|C)) → Likelihood
* (P(C)) → Prior
* (P(X)) → Evidence

---

### 🔥 “Naive” Assumption

👉 Assumes:

```text
Features are independent given the class
```

So:
$$
P(X|C) = P(x_1|C) \cdot P(x_2|C) \cdot ... \cdot P(x_n|C)
$$
👉 This simplifies computation massively

---

### 📊 Example (Spam Detection)

Features:

```text
"free", "win"
```

Compute:

```text
P(Spam | words) ∝ P(free|Spam) * P(win|Spam) * P(Spam)
```

👉 Choose class with highest probability

---

#### 🔹 1. Gaussian NB

* For continuous data
* Assumes normal distribution

---

#### 🔹 2. Multinomial NB 🔥

* For text data (word counts)
* Most common

---

#### 🔹 3. Bernoulli NB

* Binary features (0/1)

---

👉 Think like this:

1. First tree makes predictions
2. Calculate errors (residuals)
3. Next tree learns those errors
4. Repeat…

👉 Final prediction = **sum of all trees**

---

### 🔢 Core Idea (Additive Model)
$$
\hat{y} = \sum_{t=1}^{T} f_t(x)
$$
👉 Each tree (f_t) corrects previous mistakes

---

### ⚙️ Objective Function

XGBoost minimizes:

$$
\text{Obj} = \text{Loss} + \text{Regularization}
$$

* **Loss** → model error
* **Regularization** → controls complexity

---

### 🔥 Why XGBoost is Powerful

#### ✅ 1. Gradient Boosting

* Uses gradients to optimize loss

---

#### ✅ 2. Regularization

* Prevents overfitting

---

#### ✅ 3. Tree Pruning

* Removes unnecessary splits

---

#### ✅ 4. Parallel Processing

* Faster than traditional boosting

---

### ⚙️ Important Parameters (VERY IMPORTANT)

#### 🔹 n_estimators

* Number of trees

---

#### 🔹 learning_rate (eta)

* Step size
* Smaller → slower but better

---

#### 🔹 max_depth

* Tree depth
* Controls overfitting

---

#### 🔹 subsample

* Fraction of data used per tree

---

#### 🔹 colsample_bytree

* Fraction of features used

---

#### 🔹 gamma

* Minimum loss reduction to split

---

### 📊 Example Workflow

```text
Tree 1 → predicts 70
Actual → 100
Error → +30

Tree 2 → learns +30
New prediction → 90

Tree 3 → learns +10
Final → ~100
```

### 🧠 What “Regularization” means in XGBoost

👉 **Regularization = adding a penalty to make the model simpler and prevent overfitting**

In XGBoost:

> We don’t just minimize error — we also penalize **complex trees**

---

### 🔢 XGBoost Objective Function

XGBoost optimizes:
$$
Obj = Loss + \Omega(f)
$$
Where:

* **Loss** → how wrong predictions are
* **Ω(f)** → regularization term (penalty)

---

### 🔥 Regularization Term (Important)
$$
\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2
$$
👉 Let’s break this:

---

### 🔹 1. γ (gamma) — Tree Complexity Penalty

* (T) = number of leaves
* Penalizes **more splits**

👉 Effect:

* Higher γ → fewer splits → simpler tree

---

### 🔹 2. λ (lambda) — L2 Regularization

* Penalizes **large leaf weights**

👉 Effect:

* Shrinks predictions
* Similar to Ridge regression

---

### 🔹 3. α (alpha) — L1 Regularization (optional)

* Makes some weights **zero**
* Feature selection

---

### 📊 Intuition

Without regularization:

```text
Tree keeps splitting → memorizes data → overfitting ❌
```

With regularization:

```text
Tree stops splitting early → generalizes better ✅
```

---

### ⚔️ Example

#### Without regularization

```text
If age = 25 → predict 100  
If age = 26 → predict 105  
If age = 27 → predict 103
```

👉 Overfitting (too specific)

---

#### With regularization

```text
Age 20–30 → predict ~102
```

👉 Simpler & generalizable

---

### ⚙️ Key Regularization Parameters in XGBoost

| Parameter          | Meaning                 |
| ------------------ | ----------------------- |
| `gamma`            | Controls tree splitting |
| `lambda`           | L2 penalty              |
| `alpha`            | L1 penalty              |
| `max_depth`        | Limits tree size        |
| `min_child_weight` | Controls leaf creation  |

---

### 🧠 Why Regularization is Important

👉 XGBoost is very powerful → can overfit easily
👉 Regularization keeps model:

* Stable
* Generalizable
* Less noisy

---

### ⚡ Bias-Variance Perspective

* Regularization ↑ → Bias ↑ slightly
* Variance ↓ significantly

👉 Overall performance improves ✅

### ⚙️ 3. Class Weights (VERY IMPORTANT)

👉 Tell model to **penalize minority errors more**

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight='balanced')
```

👉 Works in:

* Logistic Regression
* SVM
* Random Forest
* XGBoost (`scale_pos_weight`)


### 🧠 Core Idea

👉 **Class weights = giving more importance to minority class errors**

Instead of treating all mistakes equally:

* Mistake on minority class → **high penalty**
* Mistake on majority class → **low penalty**

---

### 📊 Why Needed?

Example dataset:

| Class      | Count |
| ---------- | ----- |
| 0 (Normal) | 990   |
| 1 (Fraud)  | 10    |

👉 Model can cheat:

```text
Predict all 0 → 99% accuracy ❌
```

---
### ⚙️ 1. Class Weight (General Concept)

👉 Modify loss function:

<div style="text-align:center">
  <img src="https://raw.githubusercontent.com/sprashant433/MLOps/main/images/Part3_ROC_Curve_Screenshot.png" alt="ROC Curve Screenshot" style="max-width:100%;height:auto;" />
</div>


👉 Where:

* $w_i$ = class weight

---

### 🔥 How weights are assigned

```text
weight = total_samples / (num_classes * class_count)
```

---

#### Example:

* Total = 1000
* Class 0 = 990
* Class 1 = 10

```text
weight_0 ≈ 1000 / (2 × 990) ≈ 0.5  
weight_1 ≈ 1000 / (2 × 10) = 50
```

👉 Minority class gets **100x more importance**

---

### 🚀 Example (Logistic Regression)

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight='balanced')
model.fit(X, y)
```

👉 Internally:

* Errors on class 1 weighted more
* Model forced to learn minority patterns

---

### 🌲 2. XGBoost: `scale_pos_weight`

👉 Special parameter for imbalance

---

### 🔢 Formula

```text
scale_pos_weight = (# negative samples) / (# positive samples)
```

---

### 📊 Example

* Negative (0) = 990
* Positive (1) = 10

```text
scale_pos_weight = 990 / 10 = 99
```

---

### 🚀 Code

```python
import xgboost as xgb

model = xgb.XGBClassifier(scale_pos_weight=99)
model.fit(X, y)
```

---

### 🧠 What Happens Internally

👉 During training:

* Gradient updates for positive class are **multiplied by 99**
* Model pays more attention to minority class errors

---

### 📉 Intuition

Without weight:

```text
Model ignores minority → predicts majority
```

With weight:

```text
Minority mistakes are expensive → model learns them
```

| Class weight  | Adjust loss        |
| SMOTE         | Add synthetic data |

👉 Class weight = **no data change, only learning change**

**Metric Definitions:**

- **Accuracy:** (TP + TN) / (TP + TN + FP + FN)
- **Precision:** TP / (TP + FP) - Of predicted positives, how many are correct?
- **Recall (Sensitivity):** TP / (TP + FN) - Of actual positives, how many did we find?
- **F1-Score:** 2 * (Precision * Recall) / (Precision + Recall) - Harmonic mean
- **ROC-AUC:** Area under ROC curve - Overall performance across thresholds

**When to use which metric:**
- Balanced classes → Accuracy
- Imbalanced, cost of FP high → Precision (spam detection)
- Imbalanced, cost of FN high → Recall (disease detection)
- Balance precision/recall → F1-Score
- Overall ranking → ROC-AUC
