## 4.1 K-Means Clustering

**Concept:** Partition data into K clusters by minimizing within-cluster variance

**Steps:** 1) Initialize K centroids 2) Assign points to nearest centroid 3) Update centroids 4) Repeat

**Use Case:** Fast, works well with spherical clusters

### 📌 K-Means Clustering — Clear + Technical Explanation

K-Means is an **unsupervised learning algorithm** used to **group similar data points into K clusters**.

👉 Think like this:

* You want to group them into **K clusters**
* Each cluster has a **center (centroid)**

---

### 🚀 How It Works (Step-by-Step)

#### 1. Choose K (number of clusters)

---

#### 2. Initialize centroids randomly

---

#### 3. Assign points to nearest centroid

👉 Using distance (usually Euclidean)

---

#### 4. Update centroids

👉 Compute mean of assigned points

---

#### 5. Repeat until convergence

👉 Stop when centroids don’t change

---

### 🔢 Objective Function

K-Means minimizes:

$$
\sum_{i=1}^{K} \sum_{x \in C_i} \|x - \mu_i\|^2
$$

👉 Minimize distance between points and their cluster center

---

K = 2

👉 Clusters:

```text

```

#### 🔹 K (number of clusters)

* Most important parameter

---

#### 🔹 max_iter

* Maximum iterations

---

#### 🔹 init

* How centroids are initialized

````python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import numpy as np

# Scale features (important for K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train K-Means
kmeans = KMeans(
    n_clusters=3,           # number of clusters
    init='k-means++',       # smart initialization
    n_init=10,              # number of initializations
    max_iter=300,
    random_state=42
)
clusters = kmeans.fit_predict(X_scaled)

# Results
print(f"Inertia (within-cluster sum of squares): {kmeans.inertia_:.2f}")
print(f"Cluster centers shape: {kmeans.cluster_centers_.shape}")
print(f"Cluster distribution: {np.bincount(clusters)}")

# Visualize (2D)
if X.shape[1] >= 2:
    plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='viridis')
    plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
                s=300, c='red', marker='X', edgecolors='black', label='Centroids')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.legend()
    plt.title('K-Means Clustering')
    plt.show()
````

### Finding Optimal K (Elbow Method)

````python
# Elbow Method
inertias = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(K_range, inertias, 'bo-')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.grid()
plt.show()
# Look for "elbow" point where inertia stops decreasing sharply
````

Hierarchical Clustering is an **unsupervised learning algorithm** that builds a **tree-like structure (dendrogram)** of clusters instead of fixing K in advance.

👉 Think like this:

* Start with each point as its own cluster
* Gradually **merge closest clusters**
* Build a **tree (hierarchy)**

---

### 🚀 Types of Hierarchical Clustering

### 🔹 1. Agglomerative (Bottom-Up) 🔥 (Most Common)

```text
Each point = 1 cluster → Merge step by step → 1 big cluster
```

---

### 🔹 2. Divisive (Top-Down)

```text
Start with all points → Split recursively
```

---

### 🔁 Agglomerative Algorithm (Step-by-Step)

1. Start with N clusters (each point)
2. Compute distance between all clusters
3. Merge closest clusters
4. Repeat until 1 cluster remains

---

### 📏 Linkage Methods (VERY IMPORTANT)

👉 How distance between clusters is calculated:

---

### 🔹 1. Single Linkage

* Distance = closest points
* Can form **chains**

---

### 🔹 2. Complete Linkage

* Distance = farthest points
* Produces **compact clusters**

---

### 🔹 3. Average Linkage

* Average distance

---

### 🔹 4. Ward’s Method 🔥 (Most Used)

* Minimizes variance within clusters

---

### 🌳 Dendrogram (Key Concept)

👉 Tree structure showing merges

* Y-axis → distance
* Cut at a level → get clusters

---

### 📊 Example

Data:

```text
[1, 2, 3, 10, 11]
```

Steps:

```text
1 & 2 → merge  
→ (1,2) with 3  
→ (10,11)  
→ final merge
```

---

## 4.3 Advanced Clustering Techniques

### DBSCAN (Density-Based Spatial Clustering)

**Concept:** Clusters based on density, can find arbitrary shapes

**Pros:** No need to specify K, finds outliers

**Parameters:** eps (neighborhood radius), min_samples (core point threshold)

````python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,                # maximum distance between points in same cluster
    min_samples=5,          # minimum points to form dense region
    metric='euclidean'
)
clusters = dbscan.fit_predict(X_scaled)

# -1 indicates outliers/noise
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
n_noise = list(clusters).count(-1)
print(f"Number of clusters: {n_clusters}")
print(f"Number of noise points: {n_noise}")
print(f"Cluster distribution: {np.bincount(clusters[clusters >= 0])}")
````

### Gaussian Mixture Models (GMM)

**Concept:** Probabilistic model, assumes data from mixture of Gaussians

**Pros:** Soft clustering (probabilities), handles elliptical clusters

````python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=3,         # number of Gaussian components
    covariance_type='full', # 'full', 'tied', 'diag', 'spherical'
    random_state=42
)
clusters = gmm.fit_predict(X_scaled)
probabilities = gmm.predict_proba(X_scaled)  # soft clustering

print(f"Cluster distribution: {np.bincount(clusters)}")
print(f"Sample probabilities (first 5):\n{probabilities[:5]}")
print(f"AIC: {gmm.aic(X_scaled):.2f}")  # lower is better
print(f"BIC: {gmm.bic(X_scaled):.2f}")  # lower is better
````

## 4.4 Clustering Evaluation

**Metrics when true labels unknown (unsupervised)**

````python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# 1. Silhouette Score (-1 to 1, higher is better)
# Measures how similar point is to its cluster vs other clusters
sil_score = silhouette_score(X_scaled, clusters)
print(f"Silhouette Score: {sil_score:.3f}")
# > 0.7: Strong, 0.5-0.7: Reasonable, 0.25-0.5: Weak, < 0.25: No structure

# 2. Davies-Bouldin Index (0 to ∞, lower is better)
# Average similarity ratio of each cluster with most similar cluster
db_score = davies_bouldin_score(X_scaled, clusters)
print(f"Davies-Bouldin Index: {db_score:.3f}")

# 3. Calinski-Harabasz Index (0 to ∞, higher is better)
# Ratio of between-cluster to within-cluster dispersion
ch_score = calinski_harabasz_score(X_scaled, clusters)
print(f"Calinski-Harabasz Index: {ch_score:.2f}")

# Silhouette Plot
from sklearn.metrics import silhouette_samples
silhouette_vals = silhouette_samples(X_scaled, clusters)

plt.figure(figsize=(10, 6))
y_lower = 10
for i in range(len(set(clusters))):
    cluster_silhouette_vals = silhouette_vals[clusters == i]
    cluster_silhouette_vals.sort()
    size_cluster_i = cluster_silhouette_vals.shape[0]
    y_upper = y_lower + size_cluster_i
    plt.barh(range(y_lower, y_upper), cluster_silhouette_vals, height=1.0)
    y_lower = y_upper + 10

plt.axvline(x=sil_score, color='red', linestyle='--', label=f'Average: {sil_score:.3f}')
plt.xlabel('Silhouette Coefficient')
plt.ylabel('Cluster')
plt.title('Silhouette Plot')
plt.legend()
plt.show()
````

### 📌 Silhouette Score — How Good Are Your Clusters?

Silhouette Score is a **metric to evaluate clustering quality** — it tells you how well each point fits within its cluster **vs other clusters**.

---

### 🧠 Intuition

  <img src="images/Part4_Unsupervised_Clustering_cell16_img3_b344df98.png" alt="Image" style="max-width:90%;height:auto;" />

<div align="center">
  <img src="images/Part4_Unsupervised_Clustering_cell16_img4_4de9ecb0.png" alt="Image" style="max-width:90%;height:auto;" />

  <img src="images/Part4_Unsupervised_Clustering_cell16_img5_17de8673.png" alt="Image" style="max-width:90%;height:auto;" />
</div>

👉 For each point:

* How close it is to **its own cluster**
* How far it is from **other clusters**

---

### 🔢 Formula

$$
s = \frac{b - a}{\max(a, b)}
$$

Where:

* (a) = average distance to points in same cluster (**cohesion**)
* (b) = average distance to nearest other cluster (**separation**)

---

### 📊 Score Range

| Value  | Meaning                 |
| ------ | ----------------------- |
| **+1** | Perfect clustering ✅    |
| **0**  | Overlapping clusters ⚠️ |
| **-1** | Wrong clustering ❌      |

---

### 📊 Example

#### Case 1 (Good clustering)

```text
a = 2, b = 10
s = (10 - 2) / 10 = 0.8 ✅
```

---

#### Case 2 (Bad clustering)

```text
a = 5, b = 6
s ≈ 0.16 ❌
```

---

#### Case 3 (Wrong cluster)

```text
a = 8, b = 3
s < 0 ❌
```

---

### 🧠 Interpretation

* **a small, b large → good cluster**
* **a ≈ b → overlap**
* **a > b → misclassified**

---

### 🚀 Python Example

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

X = np.array([[1,2], [2,3], [3,4], [10,11], [11,12]])

kmeans = KMeans(n_clusters=2)
labels = kmeans.fit_predict(X)

print(score)
```

### Algorithm Comparison

| Algorithm | Need K? | Shape | Scalability | Outliers | Pros | Cons |
|-----------|---------|-------|-------------|----------|------|------|
| K-Means | Yes | Spherical | Fast | Sensitive | Simple, fast | Fixed K, spherical only |
| Hierarchical | Yes* | Any | Slow | Sensitive | Dendrogram | Slow for large data |
| DBSCAN | No | Arbitrary | Medium | Robust | Finds outliers | Sensitive to params |
| GMM | Yes | Elliptical | Medium | Moderate | Soft clustering | Assumes Gaussian |

*Can be determined from dendrogram

**Quick Decision Guide:**
- Know K, spherical clusters → K-Means
- Don't know K, arbitrary shapes → DBSCAN
- Need hierarchy → Hierarchical
- Probabilistic/soft clustering → GMM
- Outlier detection → DBSCAN
